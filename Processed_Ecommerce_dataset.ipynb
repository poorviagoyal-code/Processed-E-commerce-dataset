```python
import pandas as pd
import glob
import os

csv_files = glob.glob("*.csv")
print("CSV files available in Colab:")
for file_name in csv_files:
  print("-", file_name)

orders_data = pd.read_csv("Day9_Orders.csv")
customers_data = pd.read_csv("Day9_Customers.csv")
products_data = pd.read_csv("Day9_Products.csv")
print("All three datasets loaded successfully. ")

print("========== ORDERS DATASET ==========")
display(orders_data.head())
print("\nOrders shape:", orders_data.shape)
print("\nOrders columns:")
print(orders_data.columns.tolist())
print("\nOrders data types:")
print(orders_data.dtypes)

print("========== CUSTOMERS DATASET ==========")
display(customers_data.head())
print("\nCustomers shape:", customers_data.shape)
print("\nCustomers columns:")
print(customers_data.columns.tolist())
print("\nCustomers data types:")
print(customers_data.dtypes)

print("========== PRODUCTS DATASET ==========")
display(products_data.head())
print("\nProducts shape:", products_data.shape)
print("\nProducts columns:")
print(products_data.columns.tolist())
print("\nProducts data types:")
print(products_data.dtypes)

print("========== MISSING VALUES ==========")
print("Orders:")
print(orders_data.isnull().sum())
print("\nCustomers:")
print(customers_data.isnull().sum())
print("\nProducts:")
print(products_data.isnull().sum())

print("Orders columns:")
print(orders_data.columns.tolist())
print("\nCustomers columns:")
print(customers_data.columns.tolist())
print("\nProducts columns:")
print(products_data.columns.tolist())

orders_customers = pd.merge(
    orders_data,
    customers_data,
    on="Customer_ID",
    how="left"
)

print("Orders and customers merged successfully.")
display(orders_customers.head())

processed_data = pd.merge(
    orders_customers,
    products_data,
    on="Product_ID",
    how="left"
)

print("Orders, Customers, and Products merged successfully.")
display(processed_data.head())

print("========== PROCESSED DATASET ==========")
print("Shape:", processed_data.shape)
print("\nColumns:")
print(processed_data.columns.tolist())
display(processed_data.head(10))

first_half = orders_data.iloc[:len(orders_data) // 2]
second_half = orders_data.iloc[len(orders_data) // 2]
combined_orders = pd.concat(
    [first_half, second_half],
    ignore_index=True
)

print("Orders combined using concat()")
print("Original rows:", len(orders_data))
print("Combined rows:", len(combined_orders))
display(combined_orders.head())

processed_data["Order_Date"] = pd.to_datetime(
    processed_data["Order_Date"]
)
print("Oder_Date converetd to datetime.")
print(
    processed_data["Order_Date"].dtype
)

processed_data["Order_Month"] = (
    processed_data["Order_Date"].dt.month
)

display(
    processed_data[
        ["Order_Date", "Order_Month"]
    ].head()
)

processed_data["Order_Month_Name"] = (
    processed_data["Order_Date"].dt.month_name()
)
display(
    processed_data[
        ["Order_Date", "Order_Month_Name"]
    ].head()
)

processed_data["Order_Day"] = (
    processed_data["Order_Date"].dt.day
)

display(
    processed_data[
        ["Order_Date", "Order_Day"]
    ].head()
)

processed_data["Order_Day_Name"] = (
    processed_data["Order_Date"].dt.day_name()
)

display(
    processed_data[
        ["Order_Date", "Order_Day_Name"]
    ].head()
)

processed_data["Order_Year"] = (
    processed_data["Order_Date"].dt.year
)

display(
    processed_data[
        ["Order_Date", "Order_Day_Name"]
    ].head()
)

def classify_quantity(quantity):
  """Classify an order based on its quantity."""
  if quantity >= 5:
    return "Large"
  elif quantity >= 3:
    return "Medium"
  else:
    return "Small"

processed_data["Quantity_Category"] = (
    processed_data["Quantity"].apply(
        classify_quantity
    )
)

display(
    processed_data[
        ["Quantity", "Quantity_Category"]
    ].head(10)
)

processed_data["Calculated_Value"] = (
    processed_data.apply(
        lambda row: row["Quantity"] * row["Unit_Price"],
        axis=1
    )
)

display(
    processed_data[
        [
            "Quantity",
            "Unit_Price",
            "Calculated_Value"
        ]
    ].head()
)

print("========== PROCESSED DATASET MISSING VALUES ==========")
missing_values = processed_data.isnull().sum()
print(missing_values)

duplicate_count = processed_data.duplicated().sum()
print(f"Duplicate rows found: {duplicate_count}")
processed_data = processed_data.drop_duplicates()
print(
    f"Rows after removing duplicates: "
    f"{len(processed_data)}"
)

processed_data = processed_data.sort_values(
    by="Order_Date"
)

display(processed_data.head())

print("========== FINAL PROCESSED DATASET ==========")
print("Rows:", processed_data.shape[0])
print("Columns:", processed_data.shape[1])
print("\nColumn names:")
for column_name in processed_data.columns:
  print("-", column_name)

print("\nData types:")
print(processed_data.dtypes)

display(processed_data)

output_file = "processed_ecommerce_dataset.csv"
processed_data.to_csv(
    output_file,
    index=False
)

print(
    f"Processed dataset exported successfully "
    f"as {output_file}"
)

if os.path.exists(output_file):
  print("Export successful.")
  print(f"File created: {output_file}")
else:
  print("Exprot failed.")


```
